```python
import json
from pathlib import Path
import pandas as pd

threshold = 0.5
```

```python
def compute_stats(df):
    mean_temp = df["temperature"].mean()
    above = (df["temperature"] > threshold).sum()
    return {"mean_temp": mean_temp, "above_threshold": int(above)}
```

```python
df = pd.read_csv("examples/data_outputs/climate_readings.csv")
df["temp_f"] = df["temperature"] * 9 / 5 + 32
```

```python
df.to_csv("examples/data_outputs/derived_output.csv", index=False)
```

```python
summary = compute_stats(df)
Path("examples/data_outputs/summary.json").write_text(json.dumps(summary))
```

```python
with open("examples/data_outputs/summary.json", "r") as handle:
    raw = handle.read()
parsed = json.loads(raw)
```
